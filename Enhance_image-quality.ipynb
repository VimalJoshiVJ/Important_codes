```python
from PIL import Image

# Original image path
input_path = r"G:\Enhance image\Figure_1.jpg"

# Path where enhanced output image will be saved; {(.jpg/.tiff) in which you want to save image}
output_path = r"G:\Enhance image\Figure_1_Enhanced.tiff"

# Open original image
img = Image.open(input_path).convert("RGB")

# Display original properties
print("Original dimensions:", img.size)
print("Original DPI:", img.info.get("dpi", "Not available"))

# Increase pixel dimensions 4 times
scale_factor = 1

new_width = img.width * scale_factor
new_height = img.height * scale_factor

# High-quality upscaling using Lanczos
enhanced = img.resize(
    (new_width, new_height),
    Image.Resampling.LANCZOS
)

# Save at 1000 DPI format= "" means the format in which use are saving image (should be same as extention mentioned above)
# {quality = 100 maximum quallity, subsampling=0 colour will not reduce during compression} 
enhanced.save(
    output_path,
    format="TIFF",
    dpi=(1000, 1000),
    quality=100,     
    subsampling=0
)

# Display final properties
print("Enhancement completed!")
print("Original dimensions:", img.size)
print("New dimensions:", enhanced.size)
print("New image DPI:", Image.open(output_path).info.get("dpi"))
print("Saved at:", output_path)
```
