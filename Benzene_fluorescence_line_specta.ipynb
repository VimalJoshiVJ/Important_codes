```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import matplotlib.patches as mpatches 
```

```python
UV= pd.read_excel('Benzene_fluo DELTA.xlsx' , engine= 'openpyxl')
UV
```

```python
# Convert wavenumber (cm^-1) to wavelength (nm)
UV['Wavelength'] = 1e7 / UV['Energy']
# Plot fluorescence spectrum
#plt.figure(figsize=(8, 6))

plt.plot(UV['Wavelength'], UV['TotalSpectrum'], linewidth=2)

# Set wavelength range
plt.xlim(200, 300)
plt.ylim(bottom=(0))

# Labels
plt.xlabel('Wavelength (nm)', fontweight='bold', fontsize=13)
plt.ylabel('Intensity', fontweight='bold', fontsize=13)

# Bold tick labels
plt.xticks(fontweight='bold')
plt.yticks(fontweight='bold')

# Make axis borders thicker
for spine in plt.gca().spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')

plt.tight_layout()

# Save figure
plt.savefig('Benzene_Fluorescence_Spectrum.png', dpi=600, bbox_inches='tight')

plt.show()

```
