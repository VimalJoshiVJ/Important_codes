```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import matplotlib.patches as mpatches 
```

```python
UV= pd.read_excel('Coronene-TDDFT-B3LYP.xlsx' , engine= 'openpyxl')
```

```python
# =========================================================
# FONT SETTINGS
# =========================================================
plt.rcParams['font.family'] = 'Times New Roman'

# =========================================================
# CREATE ONE RECTANGULAR FIGURE
# =========================================================
fig, ax = plt.subplots(figsize=(8, 2.5), dpi=600)

# =========================================================
# STEM PLOT
# =========================================================
ax.stem(UV['A'],  UV['B'],  markerfmt=' ', basefmt=' ')

# =========================================================
# X-AXIS SETTINGS
# =========================================================
ax.set_xlim(170, 600)

# Tick positions:
# 170 → 200 = 30 nm
# 200 → 250 → 300 → ... = 50 nm

ax.set_xticks([170, 200, 250, 300, 350, 400, 450, 500, 550, 600])

# Hide only the 170 label
ax.set_xticklabels(['', '200', '250', '300', '350', '400', '450', '500', '550', '600'])

# =========================================================
# Y-AXIS SETTINGS
# =========================================================
ax.set_ylim(0, 1.6)

ax.set_yticks([i * 0.4 for i in range(5)])

# =========================================================
# AXIS LABELS
# =========================================================
ax.set_ylabel('Oscillator Strength', fontsize=12, fontweight='bold')

ax.set_xlabel('Wavelength (nm)', fontsize=12, fontweight='bold')

# =========================================================
# TICK LABEL SETTINGS
# =========================================================
ax.tick_params(axis='both', which='major', labelsize=11, width=2)

# Make X-axis tick labels bold

for label in ax.get_xticklabels():
    label.set_fontweight('bold')

# Make Y-axis tick labels bold

for label in ax.get_yticklabels():
    label.set_fontweight('bold')

# =========================================================
# LEGEND
# =========================================================
legend_patch = mpatches.Patch(color='white',label=r'$n$ = 6, X = -H, Y = -H')

ax.legend(handles=[legend_patch], loc='upper right', frameon=False,
    prop={
        'family': 'Times New Roman',
        'weight': 'bold',
        'size': 11
    }
)

# =========================================================
# THICK BLACK BORDER
# =========================================================
for spine in ax.spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')

# =========================================================
# LAYOUT
# =========================================================
fig.tight_layout()

# =========================================================
# SAVE HIGH-RESOLUTION FIGURE
# =========================================================
fig.savefig('Coronene_TDDFT_B3LYP.png', dpi=600, bbox_inches='tight')

# =========================================================
# SHOW
# =========================================================
plt.show()
```
