```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches 


A= pd.read_excel('Coronene_Freq_B3LYP.xlsx' , engine= 'openpyxl')


# =========================================================
# FONT SETTINGS
# =========================================================
plt.rcParams['font.family'] = 'Times New Roman'

# =========================================================
# CREATE ONE RECTANGULAR FIGURE
# =========================================================
fig, ax = plt.subplots(figsize=(8, 2.5), dpi=600)


# =========================================================
# STEM PLOT
# =========================================================
ax.stem(A['X'], A['Y'], markerfmt=' ', basefmt=' ')


# =========================================================
# X-AXIS TICK SETTINGS
# =========================================================

ax.set_xlim(0, 4000)
ax.set_xticks([500, 1000, 1500, 2000, 2500, 3000, 3500])
ax.set_xticks([0, 100, 200, 300, 400, 600, 700, 800, 900, 1100, 1200, 1300, 1400, 1600, 1700, 1800, 1900,
              2100, 2200, 2300, 2400, 2600, 2700, 2800, 2900, 3100, 3200, 3300, 3400, 3600, 3700, 3800, 
               3900, 4000], minor=True)

# Major X ticks — longer
ax.tick_params(
    axis='x',
    which='major',
    direction='out',
    length=6,
    width=1.5
)

# Minor X ticks — shorter and blank
ax.tick_params(
    axis='x',
    which='minor',
    direction='out',
    length=4,
    width=1.5
)


# =========================================================
# Y-AXIS SETTINGS
# =========================================================
ax.set_ylim(0, 110)

# Major ticks WITH NUMBERS
ax.set_yticks([0, 20, 40, 60, 80, 100])

# Minor ticks WITHOUT NUMBERS
ax.set_yticks([10, 30, 50, 70, 90], minor=True)

# Major Y ticks — longer
ax.tick_params(
    axis='y',
    which='major',
    direction='out',
    length=6,
    width=1.5
)

# Minor Y ticks — shorter and blank
ax.tick_params(
    axis='y',
    which='minor',
    direction='out',
    length=4,
    width=1.5
)

# =========================================================
# AXIS LABELS
# =========================================================
ax.set_ylabel('Intensity (km/mol)', fontsize=11, fontweight='bold')

ax.set_xlabel(r'Frequency (cm$^{-1}$)', fontsize=11, fontweight='bold')

# =========================================================
# TICK LABEL SETTINGS
# =========================================================
ax.tick_params(axis='both', which='major', labelsize=11)

# =========================================================
# MAKE X-AXIS TICK LABELS BOLD
# =========================================================
for label in ax.get_xticklabels():
    label.set_fontweight('bold')

# =========================================================
# MAKE Y-AXIS TICK LABELS BOLD
# =========================================================
for label in ax.get_yticklabels():
    label.set_fontweight('bold')

# =========================================================
# LEGEND
# =========================================================
legend_patch = mpatches.Patch(color='white', label=r'$n$ = 6; X = -H, Y = -H')

ax.legend(handles=[legend_patch], loc='upper right', frameon=False,
    prop={
        'family': 'Times New Roman',
        'weight': 'bold',
        'size': 11
    }
)

# =========================================================
# THICK BLACK BORDER
# =========================================================
for spine in ax.spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')

# =========================================================
# LAYOUT
# =========================================================
fig.tight_layout()

# =========================================================
# SAVE HIGH-RESOLUTION FIGURE
# =========================================================
fig.savefig('Coronene_Freq_B3LYP.png', dpi=600, bbox_inches='tight')

# =========================================================
# SHOW
# =========================================================
plt.show()
```
